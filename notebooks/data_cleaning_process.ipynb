# Phase 2: Data Cleaning Implementation

## Task 2.1: Handling Missing Values

```python
df['children'] = df['children'].fillna(0)
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)
```

## Task 2.2: Duplicate Detection and Removal

```python
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## Task 2.3: Outlier Detection and Treatment

```python
from scipy import stats

# Define numerical columns to analyze
numerical_cols = ['lead_time', 'adr', 'children', 'babies', 'adults', 'stays_in_weekend_nights', 'stays_in_week_nights']

# Calculate z-scores for each numerical column
z_scores = np.abs(stats.zscore(df[numerical_cols]))

# Define a threshold (e.g., z-score > 3)
threshold = 3

# Identify outliers
outliers = (z_scores > threshold).any(axis=1)
print(f"Number of outliers detected (z-score > {threshold}): {outliers.sum()}")

# Cap outliers
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
plt.figure(figsize=(15, 10))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.tight_layout()

plt.show()
```

## Task 2.4: Data Inconsistency Fixes

### Standardize categorical values

```python
df['meal'] = df['meal'].str.lower().str.strip()
```

### Handle impossible combinations

```python
# Remove records with zero guests
df = df[(df['adults'] + df['children'] + df['babies']) > 0]
```

### Fix date formats

```python
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])
```
